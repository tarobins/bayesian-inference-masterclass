# The Complete Bayesian Modeling Masterclass
### *From Classical Frequentist Estimation to Production Hamiltonian Monte Carlo*

Welcome! This 4-part hands-on curriculum provides a complete, step-by-step masterclass in estimating population parameters, diagnosing convergence, and generating predictions in **R**.

---

## The 4-Part Learning Progression

| Sheet | Title | Core Question & Key Concept | Practical Tools |
| :--- | :--- | :--- | :--- |
| **[Sheet 1](01_frequentist_vs_grid_approximation.ipynb)** | **Foundations: Frequentist vs. Grid Approx** | What is a parameter? Why do Confidence Intervals differ from Credible Intervals? How do we calculate exact posteriors on a grid? Includes **Prior Sensitivity Analysis** and **Prior Shrinkage**. | `t.test()`, `expand.grid()`, `dnorm()` |
| **[Sheet 2](02_quadratic_laplace_approximation.ipynb)** | **Rapid Prototyping: Quadratic (Laplace) Approx** | How can we approximate the posterior in 0.001 seconds using gradient optimization and the Hessian matrix? | `optim(..., hessian=TRUE)`, `rethinking::quap()` |

## What Makes This Curriculum Unique?
1. **Zero Black Boxes**: Every mathematical concept (joint likelihoods, log-sum-exp, detailed balance, Hessians) is derived from first principles with raw code.
2. **Consistent Dataset**: All 4 sheets use the same height dataset, proving that all methods converge to the exact same numbers.
3. **Active Practice**: Every sheet concludes with **Hands-On Challenge Exercises** (with expandable solutions) and **Common Pitfalls FAQs**.
